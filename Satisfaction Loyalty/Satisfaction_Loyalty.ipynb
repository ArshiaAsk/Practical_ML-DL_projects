```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv("Example.csv")
data.head()
```

```python
plt.scatter(data['Satisfaction'],data['Loyalty'])
plt.xlabel('Satisfaction')
plt.ylabel('Loyalty')
```

```python
x = data.copy()
```

```python
kmeans = KMeans(2)
kmeans.fit(x)
```

```python
cluster = x.copy()
cluster['Cluster_pred'] = kmeans.fit_predict(x)
```

```python
plt.scatter(cluster['Satisfaction'], cluster['Loyalty'], c=cluster['Cluster_pred'], cmap='rainbow')
plt.xlabel('Satisfaction')
plt.ylabel('Loyalty')
```

```python
from sklearn import preprocessing
x_scaled = preprocessing.scale(x)
x_scaled
```

```python
wcss = []
for i in range(1,10):
    kmeans = KMeans(i)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)

wcss
```

```python
plt.plot(range(1,10),wcss)
plt.xlabel('number of cluster')
plt.ylabel('WCSS')
```

```python
kmeans_new = KMeans(4)
kmeans_new.fit(x_scaled)
cluster_new = x.copy()
cluster_new['Cluster_pred'] = kmeans_new.fit_predict(x_scaled)
```

```python
cluster_new.head()
```

```python
plt.scatter(cluster_new['Satisfaction'],cluster_new['Loyalty'], c=cluster_new['Cluster_pred'], cmap='rainbow')
plt.xlabel('Satisfcation')
plt.ylabel('Loyalty')
```
